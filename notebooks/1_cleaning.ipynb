```python
import pandas as pd
from datasist.structdata import detect_outliers
import os
from pathlib import Path
```

# 2. Data Loading and Understanding

# 3. Data Cleaning

```python
# dropping customerID column as this is not required for modelling
Telco_Customer_Churn_df.drop(columns=["customerID"],inplace = True)
```

```python
# printing the unique values in all the columns
numerical_columns = ["tenure","MonthlyCharges","TotalCharges"]

for col in Telco_Customer_Churn_df.columns:
    if col not in numerical_columns:
        print(col , Telco_Customer_Churn_df[col].unique())
        print("-"*50)
```

```python
# Check Null values in the dataset
Telco_Customer_Churn_df.isnull().sum()
```

```python
# Check & drop duplicate rows if any
Telco_Customer_Churn_df.duplicated().sum()
```

```python
Telco_Customer_Churn_df.drop_duplicates(inplace=True)
```

### Handling Missing and Converting `TotalCharges`

```python
Telco_Customer_Churn_df[Telco_Customer_Churn_df["TotalCharges"] == " "]
```

```python
Telco_Customer_Churn_df["TotalCharges"] = Telco_Customer_Churn_df["TotalCharges"].replace({" ": "0.0"})
```

```python
Telco_Customer_Churn_df["TotalCharges"] = Telco_Customer_Churn_df["TotalCharges"].astype("float",errors="ignore")
```

```python
# checking the class distribution of target column
Telco_Customer_Churn_df["Churn"].value_counts()
```

**Insight:**
1. Customer ID removed as it is not required for modelling
2. No missing values in the dataset
3. Missing values in TotalCharges column were replaced with 0
4. Class imbalance identified in the target

```python
# Check outliers in numerical columns using
outlier_summary = {}

for col in numerical_columns:
    outlier_indices = detect_outliers(
        data=Telco_Customer_Churn_df,
        n=1,
        features=[col]
    )
    
    outlier_summary[col] = len(outlier_indices)
    print(f"{col}: {len(outlier_indices)} outliers")

outlier_summary
```
